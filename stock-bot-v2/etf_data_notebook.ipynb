# ETF and Stock Data Fetcher
## Raw Data from Yahoo Finance

This notebook fetches raw data from Yahoo Finance for ETFs and individual stocks without any analysis. You can easily select which symbols to fetch by editing the config.yaml file.


## 1. Setup and Imports


```python
# Import required libraries
import sys
import os
sys.path.append('.')

from etf_data_fetcher import SimpleETFDataFetcher
import pandas as pd

print("Libraries imported successfully!")
print("Ready to fetch ETF data")
```

## 2. Initialize Data Fetcher


```python
# Initialize the data fetcher with config file
fetcher = SimpleETFDataFetcher('config.yaml')

print("ETF and Stock Data Fetcher Initialized")
print(f"Tracking {len(fetcher.all_symbols)} symbols:")
print(f"  - ETFs: {len(fetcher.all_symbols)}")
print(f"  - Stocks: {len(fetcher.stock_symbols)}")

print("\nConfigured symbols:")
for symbol in fetcher.all_symbols:
    symbol_type = fetcher.symbol_types.get(symbol, 'Unknown')
    category = fetcher.symbol_categories.get(symbol, 'Unknown')
    print(f"   {symbol} [{symbol_type}]: {fetcher.symbol_names[symbol]} ({category})")
```

## 3. Show Configuration


```python
# Show current configuration
fetcher.show_config()
```

## 4. Fetch Raw Data


```python
# Fetch data using default period from config
default_period = fetcher.config['data_settings']['default_period_months']
fetcher.fetch_data(period_months=default_period)
```

## 5. Display Data Information


```python
# Display information about the fetched data
fetcher.display_data_info()
```

## SAVE IT IN DATA FOLDER

```python
fetcher.save_to_csv()
```

## 6. View Raw Data Samples


```python
# Display sample data for each ETF
for symbol in fetcher.all_symbols[:3]:  # Show first 3 ETFs
    data = fetcher.get_data(symbol)
    if data is not None:
        symbol_type = fetcher.symbol_types.get(symbol, 'Unknown')
        print(f"\n{symbol} [{symbol_type}] - {fetcher.symbol_names[symbol]}")
        print("=" * 50)
        print("First 5 rows:")
        print(data.head())
        print("\nLast 5 rows:")
        print(data.tail())
        print("\nData types:")
        print(data.dtypes)
```
